```python
import torch

def set_model_metadata(model_path, metadata):
    data = torch.load(model_path, weights_only=True, map_location='cpu')
    for key in metadata.keys():
        data[key] = metadata[key]
    torch.save(data, model_path)

set_model_metadata(
    model_path='/Users/bruno.matosak/Documents/github/flood-inference-tool/models/UNet.pt',
    metadata= {
        'use_terrain': False,
        'use_distance_map': False,
        'chip_size': 512,
        'data_preprocessing': 'normalize'
    }
)

set_model_metadata(
    model_path='/Users/bruno.matosak/Documents/github/flood-inference-tool/models/DistanceMap.pt',
    metadata= {
        'use_terrain': True,
        'use_distance_map': True,
        'chip_size': 512,
        'data_preprocessing': 'scale'
    }
)
```
