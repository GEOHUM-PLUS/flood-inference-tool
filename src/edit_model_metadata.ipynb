```python
import torch

def set_model_metadata(model_path, metadata):
    data = torch.load(model_path, weights_only=True, map_location='cpu')
    for key in metadata.keys():
        data[key] = metadata[key]
    torch.save(data, model_path)
```

```python
set_model_metadata(
    model_path='/Users/bruno.matosak/Documents/github/flood-inference-tool/models/UNet-S1.pt',
    metadata= {
        'use_terrain': False,
        'use_distance_map': False,
        'chip_size': 512,
        'data_preprocessing': 'normalize',
        'data_type': 's1_during_flood'
    }
)
```
